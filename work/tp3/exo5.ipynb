# Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# Importing the data + printing

```python
from sklearn.datasets import fetch_california_housing

cali_housing = fetch_california_housing(as_frame=True)

print(cali_housing.DESCR)

data=cali_housing.data
target=cali_housing.target
```

# Training the model

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn . model_selection import train_test_split
from sklearn. tree import plot_tree

X_train, X_test, y_train, y_test = train_test_split (data, target, test_size = 0.4 )

gini_model_overfitting = DecisionTreeRegressor(criterion='squared_error', random_state=42,max_depth=None)
gini_model_overfitting.fit(X_train,y_train)

       
```

```python
        
gini_model_pruned = DecisionTreeRegressor(criterion='squared_error', random_state=42,max_depth=7)
gini_model_pruned.fit(X_train,y_train)

```

```python
## printing the overfitted model
tree_ = gini_model_overfitting.tree_
n_nodes = tree_.node_count
print(f"Number of nodes: {n_nodes}")
# Example: Iterating through nodes to get properties
for i in range(n_nodes):
    if tree_.children_left[i] != tree_.children_right[i]:  # Internal node
        print(f"Node {i}: Split on feature {tree_.feature[i]} <= {tree_.threshold[i]:.2f}")
    else:  # Leaf node
        print(f"Node {i}: Leaf node with impurity {tree_.impurity[i]:.2f}")

```

```python
## printing the limited model
tree_ = gini_model_pruned.tree_
n_nodes = tree_.node_count
print(f"Number of nodes: {n_nodes}")
# Example: Iterating through nodes to get properties
for i in range(n_nodes):
    if tree_.children_left[i] != tree_.children_right[i]:  # Internal node
        print(f"Node {i}: Split on feature {tree_.feature[i]} <= {tree_.threshold[i]:.2f}")
    else:  # Leaf node
        print(f"Node {i}: Leaf node with impurity {tree_.impurity[i]:.2f}")
```

# Evaluation

```python
from sklearn.metrics import mean_squared_error

y_train_pred_overfitted = gini_model_overfitting.predict(X_train)
y_test_pred_overfitted = gini_model_overfitting.predict(X_test)


y_train_pred_pruned = gini_model_pruned.predict(X_train)
y_test_pred_pruned = gini_model_pruned.predict(X_test)

results = {
    'Model': ['overfitted', 'pruned'],

    'Train RMSE': [
        np.sqrt(mean_squared_error(y_train, y_train_pred_overfitted)),
        np.sqrt(mean_squared_error(y_train, y_train_pred_pruned))

    ],

    'Test RMSE': [
        np.sqrt(mean_squared_error(y_test, y_test_pred_overfitted)),
        np.sqrt(mean_squared_error(y_test, y_test_pred_pruned))
    ]
}

df_results = pd.DataFrame(results)


print(df_results)
```

we can see that the overfitted model has near 0 RMSE on the training data meaning it "memorized it", whereas the limited model has an RMSE of 0.6 which is higher, therefore indicating the model did not memorize the training data, this is demonstrated by the test RMSE, the overfitted model got a worse (higher) RMSE than the limited model, so we can deduce that limiting the depth actually helped the model be more general. 

```python
plt.figure(figsize =(200,20))
plot_tree(gini_model_pruned, filled=True, feature_names=data.columns, rounded=True, fontsize=8)
plt.show()
```
