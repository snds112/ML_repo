```r
model1 <- lm(LungCap ~ Age + Height)
summary(model1)
```

trains the model to predict lung capacity (LungCap) using Age and Height (LungCap ~ Age + Height) then shows a summary of the model information such as residuals, coefficients, and different statistical error values.

```r
cor(Age, Height, method = "pearson")
```

calculates the pearson correlation factor which shows if 2 variables are correlated, a value near 1 suggests strong correlation and a value closer to 0 means close to none, a value close to -1 shows strong negative correlation.

```r
confint(model1, coef.level = 0.95)
```

calculate a confidence interval for parameters in a fitted in the model.

```r
model2 <- lm(LungCap ~ Age + Height + Smoke + Gender + Caesarean)
summary(model2)
```

same as model1 except this one has more parameters.

```r
plot(model2)
```

similar to summary(), this function plots multiple graphs depicting various statistical data about the model and its parameters and error values.
