### Libraries:

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

### Q1:

```python
tab = pd.read_csv('DataHeight.csv',delimiter=';')
```

### Q2:

```python
x = tab . iloc [ : , 1: -1]. values
```

### Q3:

```python
y = tab . iloc [ : , -1]. values
```

### Q4:

```python
plt. scatter (x , y,color = 'blue' )
plt. title ( ' Height in function of Age ' )
plt. xlabel ( ' Age ' )
plt. ylabel ( ' Height ' )
plt. show ()
```

### Q5:
    Yes, we can see that height seems to be increasing as age increases, which indicates a possible correlation.

### Q6:

```python
pearsons_coefficient = np.corrcoef(x.flatten(), y)
print("The pearson's coeffient of the x and y inputs are: \n" ,pearsons_coefficient[0][1])
```

As the value is very close to 1 we can say that height is highly correlated with age.

### Q7:

```python
from sklearn . model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split (x, y, test_size = 0.2 )
from sklearn . linear_model import LinearRegression
regressor = LinearRegression ()
regressor.fit ( X_train , y_train)


print("Parameter Values:")
slope = regressor.coef_
intercept = regressor.intercept_
print(f"Intercept (q0): {intercept:.4f}")
print(f"Slope (q1): {slope[0]:.4f}")
```

### Q8:

```python
from sklearn.metrics import mean_squared_error

# method 1 RMSE
y_true = y_test
y_pred = regressor.predict(X_test)
mse = mean_squared_error(y_true, y_pred)
rmse=np.sqrt(mse)
print(f"Root Mean Squared Error: {rmse:.4f}")

# method 2 RSE
rse = regressor.score(x, y)
print()
print(f"(R-squared) Score/accuracy: {rse:.4f}, => error of {1-rse:.4f}")
print()
```

### Q9:

```python
print(f"Equation: y= {slope[0]:.4f}*x + {intercept:.4f} ± {rmse:.4f}")
```
